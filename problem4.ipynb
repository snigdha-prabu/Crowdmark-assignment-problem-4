If a function f is continuous on a closed interval [a, b] and f(a) · f(b) ≤ 0, then one can use the bisection method to approximate a root of f in the interval [a, b].

The idea is as follows: set L := a and R := b. If f(L) = 0 or f(R) = 0, then we have found a root and we are done. Else, we have f(L) · f(R) < 0 and so f(L) and f(R) have opposite signs.

Hence, by the intermediate value theorem, there must be a root of f in the interval (L,R).

- Now compute the midpoint c = (L + R)/2.
- If f(c) = 0 we have found an exact root and we are done.
- If |R − L| is smaller than the desired tolerance, then c is an approximate root up to the specified tolerance and we are done too.
- Else, if f(c) has the same sign as f(L), then we set L := c and.
- Else f(c) has the same sign as f(b) and we set R := c.

```python
import numpy as np

def bisection(f, a, b, max_iter=1000, tolerance=1e-10):
  #Whether an approximate root has been found
  root_approximated = False

  #Value of the approximate root (if found)
  approximate_root = None

  #Failure reason (if the algorithm fails)
  bisection_failure = False
  bisection_failure_reason = ""

  #A variable to store the number of iterations
  i = 0

  # the left and right endpoint of the interval at each step
  L = a
  R = b

  if f(L) * f(R) > 0:
    bisection_failure = True
    bisection_failure_reason = "Bisection method does not apply: f(a) and, f(b) have the same sign."
  elif f(L) == 0 or f(R) == 0:
    # an endpoint is already an exact root
    root_approximated = True
    approximate_root = L if f(L) == 0 else R
  else:
    while not root_approximated and i < max_iter:
      #implement the core logic
      i += 1
      c = (L + R) / 2
      if f(c) == 0:
        # exact root found
        root_approximated = True
        approximate_root = c
      elif abs(R - L) < tolerance:
        # interval is smaller than the tolerance, so c is an approximate root
        root_approximated = True
        approximate_root = c
      elif np.sign(f(c)) == np.sign(f(L)):
        L = c
      else:
        R = c
    if not root_approximated:
      bisection_failure = True
      bisection_failure_reason = "Did not converge within max_iter iterations."

  #print whether or not the algorithm succeeded, and either the reason why it failed, or the approximate root
  print(f"[{a}, {b}]")
  if root_approximated:
    print(f"  Success: approximate root = {approximate_root:.10f}, f(root) = {f(approximate_root):.3e}, iterations = {i}")
  else:
    print("  Failure:", bisection_failure_reason)
  print()
  return root_approximated, approximate_root, bisection_failure_reason, i


#Test cases
#Input parameters. Adapt these to test your code.
max_iter = 1000
tolerance = 1e-10

print("f(x) = e^(x/10) + ln(x/10)")
bisection(lambda x: np.exp(x/10) + np.log(x/10), 1, 10, max_iter, tolerance)

print("f(x) = arctan(x) - x^2")
bisection(lambda x: np.arctan(x) - x**2, 0, 2, max_iter, tolerance)

print("f(x) = sin(x)/ln(x)")
bisection(lambda x: np.sin(x)/np.log(x), 3, 4, max_iter, tolerance)

print("f(x) = ln(cos(x))")
bisection(lambda x: np.log(np.cos(x)), 5, 7, max_iter, tolerance)
```
